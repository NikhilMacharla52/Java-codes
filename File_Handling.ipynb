```python
# Creating and Write a File
file = open("students.txt","w")
file.write("welcome to Python Class")
file.close()
```

open() / Opening the file
"students.txt" filename
"w" write mode
write() writing information inside the file
close() closing the file

```python
# with open()
with open("students.txt","w") as file:
  file.write("Hello Students")
```

```python
with open("students.txt","r")as file:
  data = file.read()
print(data)
```

```python
#w= write, r = read, a = append - it will help to add new data to file without deleting old data
```

```python
# append
with open("notes.txt","w") as f:
  f.write("Hello Students\n")
  f.write("ML is good")
```

```python
with open("notes.txt","w") as f:
  f.write("Python\n")
```

```python
with open("notes.txt","r") as f:
  data = f.read()
print(data)
```

```python
# append
with open("notes.txt","a") as f:
  f.write("ML is good")
```

```python
with open("notes.txt","r") as f:
  data = f.read()
print(data)
print(type(data))
```

```python
# readline()
with open("notes.txt","r") as f:
  data = f.readlines()
print(data)
```

```python
# readlines()
with open("notes.txt","r") as f:
  data = f.readlines()
print(data)
print(type(data))
```

```python
#read()    return type string     Purpose "full content"
#readline()         string              1 line
#readlines           List               all lines
```

```python
# CSV

import csv
with open("marks.csv","w",newline="") as f:
  writer =csv.writer(f)
  writer.writerow(["name","marks"])
  writer.writerow(["Nick",88])
  writer.writerow(["Carry",90])
print("CSV created")
```

```python
import csv
with open("marks.csv","r") as f:
  reader = csv.reader(f)
  for row in reader:
    print(row)
```

```python
# DictReader()
with open("marks.csv","r",newline="") as f:
  reader = csv.DictReader(f)
  students = list(reader)
print(students)
print("First name:",students[0]["name"])
print("Second name:",students[1]["name"])
```
