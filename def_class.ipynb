```python
def delivery_fee(amount, fee=40):
  if amount >= 300:
    return 0
  return fee
```

```python
# MINI_PROJECT
def process_order(item,quantity,price):

  # 1. Clean item name
  item = " ".join(item.split()).title()

  # 2. Calculate subtotal
  subtotal = quantity * price

  # 3. Delivery fee
  if subtotal >= 300:
    delivery_fee = 0
  else:
    delivery_fee = 40

  # 4. Calculate total
  total = subtotal + delivery_fee

  # 5. Display
  print("Item:",item)
  print("quantity:",quantity)
  print("subtotal:",subtotal)
  print("Delivery Fee:",delivery_fee)
  print("Total:",total)

process_order("   veg hakka noodles  ",2,120)
```
